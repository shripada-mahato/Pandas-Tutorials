Print only the Product and Price columns.
Print all rows where Price > 5000.
Print the last 3 rows.
Print the column names.

```python
import pandas as pd

df = pd.DataFrame({
    "Product": ["Laptop", "Mouse", "Keyboard", "Monitor", "Printer"],
    "Price": [50000, 800, 1500, 12000, 9000],
    "Quantity": [5, 20, 10, 7, 3]
})
print(df)
print(df['Product'],df['Price'])
price = df[df['Price']>5000]
print(price)
print(df.dtypes)
print(df.info())
print(df.tail(3 ))
```
