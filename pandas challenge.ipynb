Print students whose City is Delhi.
Print students whose Age > 20 AND Marks > 80.
Print the 2nd and 3rd rows using .iloc.
Print only the Age of the student named Sneha.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "Aman", "Sneha", "Rohit", "Anjali"],
    "Age": [22, 19, 25, 21, 24, 20],
    "Marks": [85, 92, 67, 88, 75, 95],
    "City": ["Delhi", "Mumbai", "Delhi", "Kolkata", "Mumbai", "Delhi"]
})
print(df[['Name','Marks']])
print(df[df['City']=='Delhi'])
print(df[df['Marks']>=80])
print(df[(df['Age'] > 20) & (df['Marks'] > 80)])
print(df[df['Name']=='Sneha']['Age'])
```

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "Aman", "Sneha", "Rohit", "Anjali", "Karan"],
    "Age": [22, 19, 25, 21, 24, 20, 23],
    "Marks": [85, 92, 67, 88, 75, 95, 81],
    "City": ["Delhi", "Mumbai", "Delhi", "Kolkata", "Mumbai", "Delhi", "Pune"]
})
print("Original Dataset:- \n",df)
sort_marks_asscending = df.sort_values('Marks')
print("sort the marks column by asscending order:- ")
print(sort_marks_asscending)
sort_age_desscending = df.sort_values('Age',ascending=False)
print("sort the Age Column by Decending order:- \n",sort_age_desscending)
print("print top 3 student by highest marks:- \n",sort_marks_asscending.tail(3))
print("city is delhi and marks > 80:- \n",df[(df['Marks']> 80) & (df['City'] == "Delhi")])
print("only Name and marks column:- \n",df[['Name','Marks']])
```

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "Aman", "Sneha", "Rohit", "Anjali"],
    "Math": [85, 92, 67, 88, 75, 95],
    "Science": [78, 89, 72, 91, 80, 90],
    "English": [82, 94, 70, 85, 77, 93]
})
print("Original Dataset:- \n",df)
df['Total'] = (df['Math']+df['Science']+df['English'])
df['Average'] = (df['Math']+df['Science']+df['English']) / 3
print("filter dataset:- \n",df)
print("print only Name, Total, Average Column:- \n",df[['Name','Total','Average']])
print("Print students whose Average is greater than or equal to 85. \n",df[df['Average']>=85])
average_descending = df.sort_values('Average',ascending=False)
print(average_descending)
print(average_descending.head(2)[['Name','Average']])
```
