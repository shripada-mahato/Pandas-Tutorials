🐼 Challenge #9 — GroupBy (Part 1)

Use the same employee dataset.

Tasks
Find the average salary of each department.
Find the maximum salary of each department.
Find the minimum salary of each department.
Find the total salary of each department.
Count the number of employees in each department.
Find the average experience of each department.

```python
import pandas as pd

df = pd.DataFrame({
    "Emp_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Name": ["Rahul", "Priya", "Aman", "Sneha", "Rohit", "Anjali", "Karan", "Neha"],
    "Department": ["HR", "IT", "Finance", "IT", "HR", "Marketing", "Finance", "IT"],
    "City": ["Delhi", "Mumbai", "Delhi", "Kolkata", "Mumbai", "Delhi", "Pune", "Chennai"],
    "Salary": [35000, 50000, 42000, 55000, 38000, 60000, 45000, 52000],
    "Experience": [2, 5, 3, 6, 2, 7, 4, 5]
})
print(df)
ava_salary_dept = df.groupby("Department")["Salary"].mean()
max_salary_dept = df.groupby("Department")["Salary"].max()
min_salary_dept = df.groupby("Department")["Salary"].min()
total_salary_dept = df.groupby("Department")["Salary"].sum()
count_emp = df.groupby('Department')['Department'].count()
ave_exp = df.groupby('Department')['Experience'].mean()
print("Average salry of each dept:- \n",ava_salary_dept)
print("max salry of each dept:- \n",max_salary_dept)
print("min salry of each dept:- \n",min_salary_dept)
print("total salry of each dept:- \n",total_salary_dept)
print(count_emp)
print(ave_exp)
```

🐼 Challenge #10 — Aggregation (agg())

Use the same employee dataset.

Tasks
1.

Find the following statistics for Salary, grouped by Department:

Mean
Maximum
Minimum
Sum
Count

using one groupby() statement.

2.

Find the following statistics for Experience, grouped by Department:

Mean
Maximum
Minimum

using one groupby() statement.

3.

Print both results.

Rules
❌ No loops
❌ No repeated groupby() calls for the same task
✅ Use .agg().

```python
import pandas as pd

df = pd.DataFrame({
    "Emp_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Name": ["Rahul", "Priya", "Aman", "Sneha", "Rohit", "Anjali", "Karan", "Neha"],
    "Department": ["HR", "IT", "Finance", "IT", "HR", "Marketing", "Finance", "IT"],
    "City": ["Delhi", "Mumbai", "Delhi", "Kolkata", "Mumbai", "Delhi", "Pune", "Chennai"],
    "Salary": [35000, 50000, 42000, 55000, 38000, 60000, 45000, 52000],
    "Experience": [2, 5, 3, 6, 2, 7, 4, 5]
})
emp_stats = df.groupby('Department')['Salary'].agg(
    ['mean',
     'max',
     'min',
     'sum',
     'count']
)
exp_stats = df.groupby('Department')['Experience'].agg([
    'mean',
    'min',
    'max',
    'count'
])
print(emp_stats)
print(exp_stats)
```
